```python
import torch
import torch.nn as nn
```

```python
# Definição da RNN Simples
class SimpleRNN(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(SimpleRNN, self).__init__()
        self.hidden_size = hidden_size
        self.rnn_cell = nn.RNNCell(input_size, hidden_size)
    
    def forward(self, inputs):
        batch_size = inputs.shape[1]  # Formato esperado: (seq_len, batch, input_size)
        hidden = torch.zeros(batch_size, self.hidden_size)  # Estado inicial
        
        outputs = []
        for x in inputs:
            hidden = self.rnn_cell(x, hidden)  # Processa cada passo da sequência
            outputs.append(hidden)
        
        return torch.stack(outputs)  # Retorna todos os estados ocultos
```

```python
# Configurações do modelo
input_size = 1  # Tamanho da entrada (exemplo: séries temporais unidimensionais)
hidden_size = 5  # Dimensão do estado oculto
seq_length = 10  # Número de passos no tempo
batch_size = 1   # Apenas um exemplo na entrada
```

```python
# Criando a rede
rnn = SimpleRNN(input_size, hidden_size)
```

```python
# Criando entrada aleatória
inputs = torch.randn(seq_length, batch_size, input_size)

# Rodando a RNN
outputs = rnn(inputs)

# Exibindo os estados ocultos
print("Estados ocultos ao longo do tempo:")
print(outputs)
```
