```python
import os

def listar_diretorios(caminho, nivel=2, prefixo=""):
    if nivel < 0:
        return
    for item in os.listdir(caminho):
        caminho_completo = os.path.join(caminho, item)
        print(f"{prefixo}├── {item}")
        if os.path.isdir(caminho_completo):
            listar_diretorios(caminho_completo, nivel - 1, prefixo + "│   ")

# Defina o diretório raiz do projeto
raiz = "./"  # Altere para o caminho do seu projeto
listar_diretorios(raiz)
```
